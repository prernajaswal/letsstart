```
import json
import os
from tqdm import tqdm

# Paths
json_path = "/content/HRSID/HRSID_JPG/annotations/train2017.json"
images_dir = "/content/HRSID/HRSID_JPG/JPEGImages"
labels_dir = "/content/HRSID/labels/train"

os.makedirs(labels_dir, exist_ok=True)

# Load JSON
with open(json_path) as f:
    data = json.load(f)

images = {img["id"]: img for img in data["images"]}

# Convert
for ann in tqdm(data["annotations"]):
    img = images[ann["image_id"]]
    file_name = img["file_name"]
    w, h = img["width"], img["height"]

    bbox = ann["bbox"]  # [x, y, width, height]

    # Convert to YOLO
    x_center = (bbox[0] + bbox[2]/2) / w
    y_center = (bbox[1] + bbox[3]/2) / h
    bw = bbox[2] / w
    bh = bbox[3] / h

    label_path = os.path.join(labels_dir, file_name.replace(".jpg", ".txt"))

    with open(label_path, "a") as f:
        f.write(f"0 {x_center} {y_center} {bw} {bh}\n")
```

```
import os
os.makedirs("/content/HRSID/labels/val", exist_ok=True)
```

```
import json
from collections import defaultdict
from tqdm import tqdm

json_path = "/content/HRSID/HRSID_JPG/annotations/test2017.json"
labels_dir = "/content/HRSID/labels/val"

with open(json_path) as f:
    data = json.load(f)

images = {img["id"]: img for img in data["images"]}

ann_dict = defaultdict(list)
for ann in data["annotations"]:
    ann_dict[ann["image_id"]].append(ann)

for img_id, anns in tqdm(ann_dict.items()):
    img = images[img_id]
    file_name = img["file_name"]
    w, h = img["width"], img["height"]

    label_path = os.path.join(labels_dir, file_name.replace(".jpg", ".txt"))

    with open(label_path, "w") as f:
        for ann in anns:
            bbox = ann["bbox"]

            x_center = (bbox[0] + bbox[2]/2) / w
            y_center = (bbox[1] + bbox[3]/2) / h
            bw = bbox[2] / w
            bh = bbox[3] / h

            f.write(f"0 {x_center} {y_center} {bw} {bh}\n")
```
